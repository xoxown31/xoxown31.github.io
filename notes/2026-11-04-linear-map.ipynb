# Why x + 1 is not a linear map
*2026-11-04 · linear algebra*

**Question.** The graph of $S(x) = x + 1$ is a straight line. So is $S$ linear?

**What I first thought.** *[Your words.]*

**Definition.** A map $T$ is linear if, for all vectors $\mathbf{u}, \mathbf{v}$ and scalars $a, b$,
$$T(a\mathbf{u} + b\mathbf{v}) = aT(\mathbf{u}) + bT(\mathbf{v}).$$

```python
import numpy as np

A = np.array([[1, 1], [0, 1]])
def T(x): return A @ x
def S(x): return x + 1

u, v = np.array([1, 2]), np.array([3, -1])
a, b = 2, 5
```

```python
def is_linear(f):
    lhs = f(a*u + b*v)
    rhs = a*f(u) + b*f(v)
    return np.allclose(lhs, rhs)

print("T:", is_linear(T))
print("S:", is_linear(S))
```

```python
S(a*u + b*v), a*S(u) + b*S(v)
```

```python
import matplotlib.pyplot as plt

lhs, rhs = S(a*u + b*v), a*S(u) + b*S(v)
plt.figure(figsize=(6, 2.4))
plt.quiver([0, 0], [0, 0], [lhs[0], rhs[0]], [lhs[1], rhs[1]],
           color=["k", "gray"], angles="xy", scale_units="xy", scale=1)
plt.text(*lhs, "  lhs", va="top")
plt.text(*rhs, "  rhs", color="gray")
plt.xlim(-1, 30); plt.ylim(-2, 8); plt.axis("off");
```

**Why False.** Expand both sides:
$$S(a\mathbf{u} + b\mathbf{v}) = a\mathbf{u} + b\mathbf{v} + \mathbf{1}, \qquad aS(\mathbf{u}) + bS(\mathbf{v}) = a\mathbf{u} + b\mathbf{v} + (a+b)\,\mathbf{1}.$$
The left side adds $\mathbf{1}$ once, the right side $a+b = 7$ times. Gap: $6\cdot\mathbf{1}$.

Shorter test: $a = b = 0$ gives $T(\mathbf{0}) = \mathbf{0}$. A linear map fixes the origin.

```python
S(np.zeros(2))
```

So "linear" isn't about the graph being a line; only lines through the origin qualify. $x + 1$ is *affine*.

**Aside: why neural nets need ReLU.** Two linear layers collapse into one: $W_2(W_1\mathbf{x}) = (W_2W_1)\mathbf{x}$.

```python
rng = np.random.default_rng(0)
W1, W2 = rng.standard_normal((4, 3)), rng.standard_normal((2, 4))
x = rng.standard_normal(3)
relu = lambda z: np.maximum(z, 0)

one = (W2 @ W1) @ x
print(np.allclose(W2 @ (W1 @ x), one))
print(np.allclose(W2 @ relu(W1 @ x), one))
```
